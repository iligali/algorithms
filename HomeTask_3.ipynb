# Задача 1. Найти корень числа (ближайшее целое)

Нужно найти целую часть квадратного корня (округление вниз) **без** встроенных функций.

**Идея:** бинарный поиск по ответу.
- **Граница поиска:** `[0, x]`, потому что для `x ≥ 1` выполняется `sqrt(x) ≤ x`.
- **Инвариант:** ищем максимальное `mid`, для которого `mid² ≤ x`.
- **Сложность O(log x):** на каждом шаге отрезок поиска уменьшается вдвое.

```python
def my_sqrt(x):
    if x < 2:
        return x
    left, right = 0, x
    ans = 0
    while left <= right:
        mid = (left + right) // 2
        if mid * mid <= x:
            ans = mid
            left = mid + 1
        else:
            right = mid - 1
    return ans


assert my_sqrt(0) == 0
assert my_sqrt(1) == 1
assert my_sqrt(2) == 1
assert my_sqrt(4) == 2
assert my_sqrt(8) == 2
assert my_sqrt(10) == 3
assert my_sqrt(16) == 4
assert my_sqrt(26) == 5
print("все тесты прошли")
print("my_sqrt(10) =", my_sqrt(10))
```

# Задача 2. Два ксерокса

Есть `N` копий, два ксерокса: один делает копию за `x` минут, другой за `y` минут.

**Идея:** бинарный поиск по **времени** `t`.
- **Проверка достаточности:** за `t` минут первый сделает `t // x` копий, второй — `t // y`. Если сумма `≥ N` — времени хватает.
- **Верхняя граница:** `min(x, y) * N` (всё делает самый быстрый ксерокс).
- **Сложность:** O(log(min(x,y)·N)).

```python
def min_time_copies(N, x, y):
    if N == 0:
        return 0
    left, right = 0, min(x, y) * N
    while left < right:
        mid = (left + right) // 2
        if (mid // x) + (mid // y) >= N:
            right = mid
        else:
            left = mid + 1
    return left


assert min_time_copies(0, 2, 3) == 0
assert min_time_copies(1, 2, 3) == 2         
assert min_time_copies(5, 2, 3) == 6         
assert min_time_copies(4, 1, 1) == 2         
assert min_time_copies(10, 1, 1) == 5
assert min_time_copies(5, 1, 5) == 5         
print("все тесты прошли")
print("min_time_copies(5, 2, 3) =", min_time_copies(5, 2, 3))
```

# Задача 3. Накормить животных

У животных есть потребности в еде, у нас — набор неделимых порций.
Каждая порция кормит одно животное.

**Жадный алгоритм:**
1. Сортируем потребности и порции по возрастанию.
2. Идём по животным от самого «дешёвого» — подбираем наименьшую подходящую порцию.

**Почему жадно оптимально:** накормить самого неприхотливого животного минимально возможной порцией — всегда не хуже любого другого выбора: мы тратим меньше ресурса и оставляем «жирные» порции для более требовательных животных.

**Сложность:** O(n log n + m log m).

```python
def feed_animals(animals, portions):
    animals.sort()
    portions.sort()
    i = j = 0
    while i < len(animals) and j < len(portions):
        if portions[j] >= animals[i]:
            i += 1        # животное накормлено
            j += 1        # порция израсходована
        else:
            j += 1        # порция мала — пропускаем
    return i


assert feed_animals([2, 3, 4], [1, 3, 5]) == 2
assert feed_animals([1, 1, 1], [1, 1, 1]) == 3
assert feed_animals([5, 5], [1, 2, 3]) == 0
assert feed_animals([], [1, 2]) == 0
assert feed_animals([1, 2, 3, 4], [1, 2, 3, 4]) == 4
assert feed_animals([1, 3, 5, 7], [2, 4, 6, 8]) == 4   
print("все тесты прошли")
print("feed_animals([2,3,4],[1,3,5]) =", feed_animals([2, 3, 4], [1, 3, 5]))
```

# Задача 4. Найти разницу между двумя строками

Строка `b` — это `a` + 1 новый символ (порядок перемешан).

**Идея XOR:** `c ^ c = 0`, `c ^ 0 = c`. Если применить XOR ко всем символам `a` и `b`, все «парные» символы взаимно уничтожатся, останется код лишнего.

**Сложность:** O(|a| + |b|). Отлично подходит, когда важно не выделять доп. память.


```python
def find_extra_char(a, b):
    xor_sum = 0
    for ch in a:
        xor_sum ^= ord(ch)
    for ch in b:
        xor_sum ^= ord(ch)
    return chr(xor_sum)


assert find_extra_char("abcd", "abcde") == "e"
assert find_extra_char("abcd", "eabcd") == "e"
assert find_extra_char("", "x") == "x"
assert find_extra_char("a", "aa") == "a"
assert find_extra_char("abc", "bcaX") == "X"
assert find_extra_char("hello", "hellox") == "x"
print("все тесты прошли")
print("find_extra_char('abcd','abcde') =", find_extra_char("abcd", "abcde"))
```

# Задача 5. Сумма двух элементов массива (Two Sum)

Дан массив и `target`. Найти **индексы** двух элементов, дающих сумму `target`.
Каждый элемент используется один раз. Если пары нет — вернуть `[]`.

**Идея (хеш-таблица):** проходим массив один раз. Для каждого `num` проверяем, есть ли в словаре `target - num`. Если есть — нашли пару. Иначе кладём `num → i` в словарь.

**Почему это лучше двойного цикла:** мы не перебираем все пары (O(n²)), а за O(1) в среднем проверяем наличие дополнения. Итого O(n) по времени и O(n) по памяти.

```python
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        diff = target - num
        if diff in seen:
            return [seen[diff], i]
        seen[num] = i
    return []


assert two_sum([2, 7, 11, 15], 9) == [0, 1]
assert two_sum([3, 2, 4], 6) == [1, 2]
assert two_sum([3, 3], 6) == [0, 1]
assert two_sum([1, 2, 3], 7) == []
assert two_sum([-1, -2, -3, -4, -5], -8) == [2, 4]   
assert two_sum([0, 4, 3, 0], 0) == [0, 3]
print("все тесты прошли")
print("two_sum([2,7,11,15], 9) =", two_sum([2, 7, 11, 15], 9))
```

# Задача 6. Массив анаграмм

Сгруппировать строки-анаграммы вместе. Порядок групп и элементов внутри — любой.

**Идея:** анаграммы дают одинаковый набор символов. Используем **отсортированную строку как ключ** словаря (тоже самое — `tuple(sorted(s))`).

**Сложность:** O(n · k log k), где `n` — число строк, `k` — макс. длина. Для коротких строк это очень быстро.


```python
from collections import defaultdict

def group_anagrams(strs):
    groups = defaultdict(list)
    for s in strs:
        key = tuple(sorted(s))   # отсортированный набор символов — ключ
        groups[key].append(s)
    return list(groups.values())


# Чтобы assert корректно отрабатывали
def _normalize(groups):
    return sorted([sorted(g) for g in groups])

res = group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"])
expected = [["bat"], ["nat", "tan"], ["ate", "eat", "tea"]]
assert _normalize(res) == _normalize(expected)

assert group_anagrams([]) == []
assert _normalize(group_anagrams([""])) == [[""]]
assert _normalize(group_anagrams(["a"])) == [["a"]]
assert _normalize(group_anagrams(["abc", "cba", "bac", "xyz", "zyx"])) \
       == _normalize([["abc", "cba", "bac"], ["xyz", "zyx"]])

print("все тесты прошли")
print("group_anagrams(...) =", res)
```
