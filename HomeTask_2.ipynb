```python
class Node:
    __slots__ = ("data", "next")

    def __init__(self, data, next=None):
        self.data = data
        self.next = next


class OneDirectionList:
    def __init__(self, head=None):
        self.head = head

    @classmethod
    def from_iterable(cls, values):
        obj = cls()
        tail = None
        for value in values:
            node = Node(value)
            if obj.head is None:
                obj.head = node
            else:
                tail.next = node
            tail = node
        return obj


def to_values(lst):
    result = []
    cur = lst.head
    while cur:
        result.append(cur.data)
        cur = cur.next
    return result
```

# Развернуть односвязный список

```python
def reverse_list(lst):
    prev = None
    curr = lst.head
    while curr:
        nxt = curr.next
        curr.next = prev
        prev = curr
        curr = nxt
    lst.head = prev
    return lst


lst = OneDirectionList.from_iterable([1, 2, 3, 4])
assert to_values(reverse_list(lst)) == [4, 3, 2, 1]

lst = OneDirectionList.from_iterable([])
assert to_values(reverse_list(lst)) == []

lst = OneDirectionList.from_iterable([7])
assert to_values(reverse_list(lst)) == [7]

print('все тесты прошли успешно')
```

# Найти середину списка

```python
def middle_node(lst):
    slow = fast = lst.head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow


lst = OneDirectionList.from_iterable([1, 2, 3, 4, 5])
mid = middle_node(lst)
assert mid is not None and mid.data == 3

lst = OneDirectionList.from_iterable([1, 2, 3, 4])
mid = middle_node(lst)
assert mid is not None and mid.data == 3

lst = OneDirectionList.from_iterable([])
assert middle_node(lst) is None

print('все тесты прошли успешно')
```

# Удалить элемент из односвязного списка

```python
def delete_node(lst, val):
    while lst.head and lst.head.data == val:
        lst.head = lst.head.next

    curr = lst.head
    while curr and curr.next:
        if curr.next.data == val:
            curr.next = curr.next.next
            break
        curr = curr.next
    return lst


lst = OneDirectionList.from_iterable([1, 2, 3, 2, 4])
assert to_values(delete_node(lst, 2)) == [1, 3, 2, 4]

lst = OneDirectionList.from_iterable([1, 2, 3])
assert to_values(delete_node(lst, 1)) == [2, 3]

lst = OneDirectionList.from_iterable([1, 2, 3])
assert to_values(delete_node(lst, 5)) == [1, 2, 3]

print('все тесты прошли успешно')
```

# Является ли одна строка исходной для другой

```python
def is_subsequence(a, b):
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] == b[j]:
            i += 1
        j += 1
    return i == len(a)


assert is_subsequence("abc", "a1b2c3") is True
assert is_subsequence("abc", "acb") is False
assert is_subsequence("", "xyz") is True
assert is_subsequence("a", "") is False

print('все тесты прошли успешно')
```

# Поиск пары с суммой target

```python
def has_pair_with_sum(arr, target):
    left, right = 0, len(arr) - 1
    while left < right:
        s = arr[left] + arr[right]
        if s == target:
            return True
        elif s < target:
            left += 1
        else:
            right -= 1
    return False


assert has_pair_with_sum([1, 2, 3, 4, 6], 6) is True
assert has_pair_with_sum([1, 2, 3, 4, 6], 20) is False
assert has_pair_with_sum([], 5) is False
assert has_pair_with_sum([2, 2], 4) is True

print('все тесты прошли успешно')
```

# Является ли слово палиндромом

```python
def is_palindrome(s):
    left, right = 0, len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
    return True


assert is_palindrome("level") is True
assert is_palindrome("hello") is False
assert is_palindrome("") is True
assert is_palindrome("abba") is True

print('все тесты прошли успешно')
```

# Удаление дубликатов из отсортированного массива

```python
def remove_duplicates(arr):
    if not arr:
        return 0

    slow = 0
    for fast in range(1, len(arr)):
        if arr[fast] != arr[slow]:
            slow += 1
            arr[slow] = arr[fast]
    return slow + 1


arr = [1, 1, 2, 2, 3]
n = remove_duplicates(arr)
assert n == 3
assert arr[:n] == [1, 2, 3]

arr = [1, 1, 1]
n = remove_duplicates(arr)
assert n == 1
assert arr[:n] == [1]

arr = []
assert remove_duplicates(arr) == 0

print('все тесты прошли успешно')
```

# Слияние двух отсортированных списков

```python
def merge_sorted_lists(lst1, lst2):
    dummy = Node(0)
    tail = dummy

    a = lst1.head
    b = lst2.head
    while a and b:
        if a.data <= b.data:
            tail.next = a
            a = a.next
        else:
            tail.next = b
            b = b.next
        tail = tail.next

    tail.next = a if a else b
    return OneDirectionList(head=dummy.next)


lst1 = OneDirectionList.from_iterable([3, 6, 8])
lst2 = OneDirectionList.from_iterable([4, 7, 9, 11])
merged = merge_sorted_lists(lst1, lst2)
assert to_values(merged) == [3, 4, 6, 7, 8, 9, 11]

merged = merge_sorted_lists(
    OneDirectionList.from_iterable([]),
    OneDirectionList.from_iterable([1, 2]),
)
assert to_values(merged) == [1, 2]

merged = merge_sorted_lists(
    OneDirectionList.from_iterable([1, 2]),
    OneDirectionList.from_iterable([]),
)
assert to_values(merged) == [1, 2]

merged = merge_sorted_lists(
    OneDirectionList.from_iterable([]),
    OneDirectionList.from_iterable([]),
)
assert to_values(merged) == []

print('все тесты прошли успешно')
```
