## 1. Наибольшая непрерывная возрастающая последовательность

При строгом росте увеличиваем текущую длину, иначе начинаем с 1. После каждого элемента обновляем максимум. Время - $O(n)$, память - $O(1)$.

```python
def longest_run(nums):
    if not nums:
        return 0

    current = best = 1
    for i in range(1, len(nums)):
        if nums[i] > nums[i - 1]:
            current += 1
        else:
            current = 1
        best = max(best, current)
    return best


for nums, expected in [
    ([3, 2, 8, 9, 5, 10], 3),
    ([1, 2, 7, 9, 0, 10], 4),
    ([8, 8, 8, 8], 1),
    ([4, 1, 2, 3], 3),
    ([], 0),
]:
    result = longest_run(nums)
    assert result == expected
    print(nums, "->", result)
```

## 2. Треугольник Паскаля

По краям строки ставим единицы. Каждое число внутри - сумма двух соседних чисел из предыдущей строки. Время и память для $n$ строк - $O(n^2)$.

```python
def pascal_triangle(n):
    triangle = []
    for i in range(n):
        row = [1] * (i + 1)
        for j in range(1, i):
            row[j] = triangle[i - 1][j - 1] + triangle[i - 1][j]
        triangle.append(row)
    return triangle


assert pascal_triangle(0) == []
assert pascal_triangle(1) == [[1]]

triangle = pascal_triangle(5)
assert triangle == [[1], [1, 1], [1, 2, 1], [1, 3, 3, 1], [1, 4, 6, 4, 1]]
for row in triangle:
    print(*row)
```

## 3. Размен монет

Если последняя монета имеет номинал $c$, для остатка $s-c$ уже известно минимальное число монет. Перебираем все подходящие номиналы и выбираем минимум из $dp[s-c]+1$. Для суммы $A$ и $k$ номиналов время - $O(Ak)$, память - $O(A)$.

```python
def min_coins(coins, amount):
    limit = amount + 1
    dp = [limit] * (amount + 1)
    dp[0] = 0

    for s in range(1, amount + 1):
        for coin in coins:
            if coin <= s:
                dp[s] = min(dp[s], dp[s - coin] + 1)

    if dp[amount] == limit:
        return -1
    return dp[amount]


for coins, amount, expected in [
    ([1, 3, 4], 6, 2),
    ([2], 3, -1),
    ([5, 2], 0, 0),
    ([], 7, -1),
]:
    result = min_coins(coins, amount)
    assert result == expected
    print(f"Номиналы {coins}, сумма {amount} -> {result}")
```

## 4. Максимальный палиндром в строке

Перебираем центры на символах и между ними. Расширяем границы, пока символы совпадают, и запоминаем самый длинный палиндром. Время - $O(n^2)$, дополнительная память - $O(1)$.

```python
def longest_palindrome(s):
    best_start = best_len = 0
    for i in range(len(s)):
        for right in (i, i + 1):
            left = i
            while left >= 0 and right < len(s) and s[left] == s[right]:
                left -= 1
                right += 1

            length = right - left - 1
            if length > best_len:
                best_start = left + 1
                best_len = length

    return s[best_start:best_start + best_len]


for s, expected in [
    ("babad", "bab"),
    ("cbbd", "bb"),
    ("топот", "топот"),
    ("aaaa", "aaaa"),
    ("abc", "a"),
    ("", ""),
]:
    result = longest_palindrome(s)
    assert result == expected
    print(f"{s!r} -> {result!r}")
```
