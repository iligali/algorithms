# Сумма чисел: 1, ... , 1000000

```python
def sum_N(first, last):
    sum = 0
    while(first <= last):
        sum += first + last
        if first == last:
            sum -= first
            return sum
        first += 1
        last -= 1
    return sum

print(f"1; sum = {sum_N(1, 1)}")
print(f"1 + 2; sum = {sum_N(1, 2)}")
print(f"1 + 2 + 3; sum = {sum_N(1, 3)}")
print(f"1 + 2 + 3 + 4; sum = {sum_N(1, 4)}")
print(f"1 + 2 + 3 + 4 + 5; sum = {sum_N(1, 5)}")
print("...")
print(f"1 + ... + 1000000; sum = {sum_N(1, 1000000)}")
```

# Two sum
- Дан отсортированный массив по возрастанию: arr = (a_{1}, a_{2}, ..., a_{k-1}, a_{k}, ... , a_{n}) и target - целое число

- Необходимо в массиве найти a_i , a_j : target = a_i + a_j

```python
def two_sum(arr, target):
    n = len(arr)
    left = 0
    right = n - 1
    while (left < right):
        s = arr[left] + arr[right]
        if s < target:
            left += 1
        elif s > target:
            right -= 1
        else:
            return arr[left], arr[right]
    return -1, -1

print(f"arr = [1, 2], target = 3, answer = {two_sum([1, 2], 3)}")
print(f"arr = [1, 2, 3], target = 5, answer = {two_sum([1, 2, 3], 5)}")
print(f"arr = [1, 2, 3, 4], target = 7, answer = {two_sum([1, 2, 3, 4], 7)}")
print(f"arr = [1, 2, 3, 4], target = 8, answer = {two_sum([1, 2, 3, 4], 8)}")
        
        
```

# Инверсия массива
- input: массив целых чисел

- задача: необходимо его развернуть(Пример: [1, 2, 3] -> [3, 2, 1])

- требование: O(n), без доп.памяти

```python
def arr_inversion(arr):
    left = 0
    right = len(arr) - 1
    while left <= right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return arr

print(f"{[1, 2, 3]} -> {arr_inversion([1, 2, 3])}")
print(f"{[1, 7, 3, 10]} -> {arr_inversion([1, 7, 3, 10])}")
```

# Сдвиг части массива

- input: массив целых чисел, величина сдвига

- задача: необходимо сдвинуть(на величину сдвига) справа налево часть массива

- пример: arr = (1,2,3,4,5,6,7); k=3 -> (5,6,7,1,2,3,4)

```python
1, 2, 3, 4, 5, 6, 7
0, 1, 2, 3, 4, 5, 6
k=3

def arr_inversion(arr):
    left = 0
    right = len(arr) - 1
    while left <= right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return arr

def arr_offset(arr, k):
    n = len(arr)
    if k < 0:
        raise ValueError("k - положительное целое число")
    if k == 0:
        return arr
    if k > n:
        k = k % n
    arr = arr_inversion(arr) # O(n)
    arr[:k] = arr_inversion(arr[:k]) # O(k)
    arr[k:] = arr_inversion(arr[k:]) # O(k)
    return arr


print(f"k=3; {[1,2,3,4,5,6,7]} -> {arr_offset([1,2,3,4,5,6,7], 3)}")
print(f"k=0; {[1,2,3,4,5,6,7]} -> {arr_offset([1,2,3,4,5,6,7], 0)}")
#print(f"k=-1; {[1,2,3,4,5,6,7]} -> {arr_offset([1,2,3,4,5,6,7], -1)}")
print(f"k=9; {[1,2,3,4,5,6,7]} -> {arr_offset([1,2,3,4,5,6,7], 9)}")

    
```

# Слияние двух отсортированных массивов(с доп.aллокацией)

- input: arr1, arr2 - отсортированные массивы 

- задача: надо объединить arr1, arr2 в arr3 - отсортированный массив

```python
def merge(arr1, arr2):
    n1, n2 = len(arr1), len(arr2)
    i1, i2, i3 = 0, 0, 0
    arr3 = [0]*(n1 + n2) # выделяем память под 3 массив
    while(i1 < n1):
        if (i2 >= n2):
            arr3[i3:] = arr1[i1:]
            return arr3
        if (arr1[i1] <= arr2[i2]):
            arr3[i3] = arr1[i1]
            i1 += 1
            i3 += 1
        else:
            arr3[i3] = arr2[i2]
            i2 += 1
            i3 += 1
    if (i3 < n1 + n2):
        arr3[i3:] = arr2[i2:]
    return arr3

print(f"[1, 3], [1, 2, 4, 5] -> {merge([1, 3], [1, 2, 4, 5])}")
print(f"[1, 3, 7, 9, 11], [1, 2, 4] -> {merge([1, 3, 7, 9, 11], [1, 2, 4])}")   
print(f"[], [1, 2, 4] -> {merge([], [1, 2, 4])}")
print(f"[1, 3, 7, 9, 11], [] -> {merge([1, 3, 7, 9, 11], [])}")   
        
    
```

# Слияние двух массивов(без аллокации)

```python

def merge_without(arr1, arr2):
    n1, n2 = len(arr1), len(arr2)
    i1 = n1 - n2 - 1
    i2 = n2 - 1
    control = n1 - 1
    while (i1 >= 0):
        if i2 < 0:
            return arr1
        if arr1[i1] >= arr2[i2]:
            arr1[control] = arr1[i1]
            control -= 1
            i1 -= 1
        else: 
            arr1[control] = arr2[i2]
            control -= 1
            i2 -= 1
    return arr1

print(f"[1, 3, 5, 0, 0], [5, 7] -> {merge_without([1, 3, 5, 0, 0], [5, 7])}")
print(f"[1, 3, 5, 0, 0], [1, 3] -> {merge_without([1, 3, 5, 0, 0], [1, 3])}")
print(f"[1, 3, 5, 0, 0], [3, 7] -> {merge_without([1, 3, 5, 0, 0], [3, 7])}")
print(f"[1, 3, 5], [] -> {merge_without([1, 3, 5], [])}")
        
    
    
```

# Минимальный размер

- input: массив положительных чисел(nums), положительное число(target)

- задача: верните подмассив минимальной длины: сумма элементов >= target



```python
def min_size(nums, target):
    n = len(nums)
    begin, end = 0, 0
    ans_begin, ans_end = 0, 0
    s = nums[begin]
    while (end < n):
        if (s >= target):
            ans_begin, ans_end = begin, end
            s -= nums[begin]
            begin += 1
        else: 
            end += 1
            if end >= n:
                continue
            s += nums[end]
    return nums[ans_begin : ans_end + 1]

print(f"target=7, nums=[2,3,1,2,4,3] -> {min_size([2,3,1,2,4,3], 7)}")
print(f"target=5, nums=[2,3,1,2,4,3] -> {min_size([2,3,1,2,4,3], 5)}")
print(f"target=4, nums=[2,3,1,2,4,3] -> {min_size([2,3,1,2,4,3], 4)}")  
            
```

# Сортировка массива из 0 и 1

- input: массив из 0, 1

- задача: отсортировать массив по возрастанию

```python
def sort_zero_one(arr):
    n = len(arr)
    left = 0
    right = n - 1
    while (left <= right):
        if (arr[left] == 1) & (arr[right] == 0):
            arr[left], arr[right] = arr[right], arr[left]
            left += 1
            right -= 1
        elif (arr[left] == 0) & (arr[right] == 1):
            left += 1
            right -= 1
        elif (arr[left] == 0) & (arr[right] == 0):
            left += 1
        elif (arr[left] == 1) & (arr[right] == 1):
            right -= 1
    return arr

print(f"[0, 1, 1, 0] -> {sort_zero_one([0, 1, 1, 0])}")
print(f"[0, 1, 1, 1, 0] -> {sort_zero_one([0, 1, 1, 1, 0])}")   
print(f"[1, 1, 1, 0] -> {sort_zero_one([1, 1, 1, 0])}")        
            

        
    
```

# Задача о флаге Нидерландов

- input: массив из 0, 1, 2

- задача: требуется отсортировать его за линейное время

```python
def sort_012(arr):
    n = len(arr)
    hashes = 3 * [0] # Выделение памяти для массива хешей
    for i in range(n): # O(n)
        if arr[i] == 0:
            hashes[0] += 1
        elif arr[i] == 1:
            hashes[1] += 1
        else:
            hashes[2] += 1
            
    for i in range(0, hashes[0]):
        arr[i] = 0
    
    for i in range(hashes[0], hashes[0] + hashes[1]):
        arr[i] = 1
        
    for i in range(hashes[0] + hashes[1], hashes[0] + hashes[1] + hashes[2]):
        arr[i] = 2
        
    return arr

print(f"[0, 2, 1, 1, 0, 2] -> {sort_012([0, 2, 1, 1, 0, 2])}")
print(f"[0, 2, 1, 1] -> {sort_012([0, 2, 1, 1])}")
print(f"[0, 1, 1] -> {sort_012([0, 1, 1])}")
print(f"[2, 1, 1] -> {sort_012([2, 1, 1])}")   
```

# Четные числа вперед

- input: не отсортированный массив целых чисел

- задача: необходимо перенести четные числа в начало, сохранив порядок

```python
def forward_even(nums):
    n = len(nums)
    control, search = 0, 0
    while (search < n):
        if (nums[search] % 2 == 0):
            nums[control], nums[search] = nums[search], nums[control]
            search += 1
            control += 1
        else:
            search += 1
    return nums

print(f"[1, 3, 5, 2, 4] -> {forward_even([1, 3, 5, 2, 4])}")
print(f"[2, 4, 4, 3] -> {forward_even([2, 4, 4, 3])}")
print(f"[10, 7, 9, 5, 3, 2] -> {forward_even([10, 7, 9, 5, 3, 2])}")
```

# Нули в конец

- input: массив, который включает 0
- задача: перекинуть все нули в конец

```python
def zeros_to_end(nums):
    n = len(nums)
    search = 0
    place = n - 1
    while (search < place):
        if (nums[search] == 0):
            nums[place], nums[search] = nums[search], nums[place]
            search += 1
            place -= 1
        else:
            search += 1
    return nums

print(f"[0, 1, 2, 0, 3] -> {zeros_to_end([0, 1, 2, 0, 3])}")
print(f"[0, 0, 3] -> {zeros_to_end([0, 0, 3])}")
print(f"[3] -> {zeros_to_end([3])}")
```
